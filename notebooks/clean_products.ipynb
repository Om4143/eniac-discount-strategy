Check! Is there any pattern in corrupted promo_price w.r.t. price 

```python
# Compare the badly-formatted promo_price values against the (already fixed) price column.
numeric_attemp = pd.to_numeric(df_product['promo_price'], errors = 'coerce')
df_product[['promo_price', 'price']][numeric_attemp.isna()].head(20)
```

Function to shift first decimal to right 2 digits

```python
# Write a function that fixes prices with an extra decimal point, then apply it to promo_price.
def fix_promo_price(s):
    if pd.isna(s):
        return None
    s = str(s)
    dots = s.count('.')

    if dots == 2:
        first_dot_pos = s.index('.')
        digits = s.replace('.', '')
        new_pos = first_dot_pos + 2
        return round(float(digits[:new_pos] + '.' + digits[new_pos:]), 2)
    elif dots == 1:
        return round(float(s) / 10, 2)
    else:
        return round(float(s), 2)

 
df_product['promo_price_fixed'] = df_product['promo_price'].apply(fix_promo_price)
```
